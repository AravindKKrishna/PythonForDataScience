```python
import numpy as np
import pandas as pd
array = np.array([[2600,55000],[3000,56500],[3200,61000],[3600,68000],[4000,72500]])
df=pd.DataFrame(data=array,index=[1,2,3,4,5],columns=["size","price"])
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(3,2))
sns.scatterplot(x=df["size"],y=df["price"])
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
# Create linear regression object
model = LinearRegression()
```

```python
# Create and fit the model
train_model = model.fit(df[["size"]], df["price"])#X=train data, y=target variable
```

```python
# Predict using the model
y_pred = model.predict([[3200]])
```

```python
print("house price for the size of 3200 is",y_pred)
```

```python
m=model.coef_# to find the weights
print (m)
```

```python
c=model.intercept_
print(c)
```

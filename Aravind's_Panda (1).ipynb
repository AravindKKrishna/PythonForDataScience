```python

def sort_dataframe(df):
     return df.sort_values(by=df.columns.tolist(), ascending=True)


# Sample DataFrame
data = {
    'Name': ['David', 'Alice', 'Charlie'],
    'Age': [25, 22, 23]
}

df = pd.DataFrame(data)

# Call the function
sorted_df = sort_dataframe(df)

# Display the sorted DataFrame
print(sorted_df)
```

```python

data = {
    'Item': ['Pen', 'Notebook', 'Eraser'],
    'Quantity': [10, 5, 3],
    'Price': [5, 20, 2]
}

df = pd.DataFrame(data)
df
```

```python
def totalColumn(df):
    df["Total"]=df["Quantity"]*df["Price"]
    return df
```

```python

df=totalColumn(df)
print(df)
```

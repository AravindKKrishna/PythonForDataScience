```python
a={
   'sepal_length' : [5.1,4.9,4.7,3.0,6.8,6.7,6.5,6.2,5.9], 
   'sepal_width'  : [3.5,3.0,3.2,4.4,2.8,3.0,3.0,3.4,3.0] 
}
b=pd.DataFrame(a)
print(b)
```

```python
sns.scatterplot(b,x="sepal_length",y="sepal_width")

```

```python
from sklearn.cluster import KMeans
```

```python
kmeans=KMeans(n_clusters=2)
```

```python


# Predict the cluster for each data point
labels = kmeans.fit_predict(b)
```

```python
labels
```

```python
b["cluster"]=labels
b
```

```python
sns.scatterplot(b,x="sepal_length",y="sepal_width",hue="cluster")
```

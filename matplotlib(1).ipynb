```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1,2,3],[2,4,6]) #ploting line using X and Y coordinates
plt.show()
```

```python
plt.figure(figsize=[10,5])
plt.plot([4,6,8,9]) #plot line graph only use Y axis, defult X value is 0,0.5,1,1.5....
plt.show()
```

```python
plt.figure(figsize=(4, 4), dpi=100) #Clean, separate canvases, then plt.figure() sets up a board to draw it on.
plt.plot([2,4,1,5])                 #  plt.plot() draws a chart
plt.show()
```

```python
# The other arguments in plot() is "Marker","Line type", "Coulor"
plt.figure(figsize=[4,5])
plt.plot([2,4,1,5],c='r')  #r for red, b for blue, g for green               #
plt.show()
```

```python
plt.figure(figsize=[4,5])
plt.plot([2,4,1,5],c='r',ls='--')  # the argument ls or line styl is used of stylize the line
```

```python
plt.figure(figsize=[4,5])
plt.plot([2,4,1,5],c='r',ls='dotted')
```

```python
#multiple line
#ploting X and Y axis separetly
X=[3,5,7,9]
Y=[2,4,6,10]

plt.plot(X,ls="dotted")
plt.plot(Y, ls="solid",marker='P')

plt.show()
```

```python
# Marker argument gives data points on the line

plt.plot(X,ls="dotted", marker='o', mfc='r')
plt.plot(Y, ls="solid", marker='P',markersize=20, markeredgecolor='g') # markersize or ms, markeredgecolor or mec, markerfacecolor or mfc

plt.show()
```

```python
#Marker + Line + color

plt.plot(X, '*:r')

plt.show()
```

```python
#  ploting grid line

plt.figure(figsize=[4,5])
plt.grid(color='g')
```

```python
# subplot() drown multiple plot in one figure

#syntax: subplot(rows, columns, index)


plt.figure(figsize=[5,5])

plt.subplot(2,1,1)
plt.subplot(2,1,2)

plt.show() # 2 rows and 1 column
```

```python
plt.figure(figsize=[5,5])


plt.subplot(2,2,3)
plt.subplot(2,2,2)

plt.show() # 1 row and 2 columns
```

```python
plt.figure(figsize=[5,5])


plt.subplot(223)
plt.subplot(222)

plt.show() # 1 row and 2 columns
```

```python
plt.figure(figsize=[5,5])

plt.subplot(2,2,1)
plt.title("Plot 1")
plt.plot([1, 2, 3], [1, 4, 9])

plt.subplot(2,2,2)
plt.title("Plot 2")
plt.plot([1, 2, 3], [1, 2, 3])

plt.subplot(2,2,3)
plt.title("Plot 3")
plt.plot([1, 2, 3], [3, 2, 1])

plt.subplot(2,2,4)
plt.title("Plot 4")
plt.plot([1, 2, 3], [9, 4, 1])

plt.show() # 2 rows and 2 columns
```

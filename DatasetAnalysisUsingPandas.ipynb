```python
import matplotlib.pyplot as plt
```

```python
ax=plt.axes(projection="3d")
ax.plot(df["Price_range"],df["Votes"],df["Average_Cost_for_two"])
ax.set_xlabel("PriceRange")
ax.set_ylabel("votes")
ax.set_zlabel("avgCost")
ax.set_title("#D visualization of three zomato attributes")
plt.show()
```

```python

plt.scatter(df["Price_range"],df["Votes"], color='red', marker='o')
plt.title("Scatter Plot Example")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

```python
plt.bar(df["Price_range"],df["Votes"], color='green')
plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

```python
# Multiple Plots in One Figure
```

```python
x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]

plt.subplot(1, 2, 1)  # 1 row, 2 columns, 1st plot
plt.plot(x, y1, 'r-')
plt.title("Square Numbers")

plt.subplot(1, 2, 2)  # 1 row, 2 columns, 2nd plot
plt.plot(x, y2, 'b-')
plt.title("Linear Numbers")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y1 = [1, 4, 9, 16]
y2 = [1, 2, 3, 4]
y3 = [4, 3, 2, 1]

plt.subplot(3, 1, 1)  # 3 rows, 1 column, first plot
plt.plot(x, y1, 'r-')
plt.title("Plot 1")

plt.subplot(3, 1, 2)  # second plot
plt.plot(x, y2, 'g-')
plt.title("Plot 2")

plt.subplot(3, 1, 3)  # third plot
plt.plot(x, y3, 'b-')
plt.title("Plot 3")

plt.tight_layout()  # fix overlapping
plt.show()
```

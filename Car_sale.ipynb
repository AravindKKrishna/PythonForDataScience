```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Step 1: Clean column names (you've already done this)
a.columns = a.columns.str.strip()

# Step 2: Define target and features
target = 'Price_in_thousands'
features = [
    '__year_resale_value', 'Engine_size', 'Horsepower', 'Width',
    'Curb_weight', 'Fuel_capacity', 'Fuel_efficiency', 'Power_perf_factor'
]

X = a[features]
y = a[target]

# Step 3: Train the model
model = LinearRegression()
model.fit(X, y)

# Step 4: Make predictions
y_pred = model.predict(X)


# Step 5: Evaluate the model
r2 = r2_score(y, y_pred)
mse = mean_squared_error(y, y_pred)

print("R² Score:", round(r2, 4))
print("Mean Squared Error:", round(mse, 2))

# Optional: View coefficients
coefficients = pd.Series(model.coef_, index=features)
print("\nFeature Coefficients:\n", coefficients.sort_values(ascending=False))
```

```python
# Predict prices for all cars in the dataset
y_pred = model.predict(X)

# Add predictions to the DataFrame
a['Predicted_price'] = y_pred

# Preview actual vs predicted
print(a[[target, 'Predicted_price']].head())
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Step 1: Clean column names
a.columns = a.columns.str.strip()

# Step 2: Select features and target
features = ['__year_resale_value', 'Engine_size', 'Horsepower', 'Width', 'Curb_weight']
target = 'Price_in_thousands'

X = a[features]
y = a[target]

# Step 3: Scale features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Split data
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# Step 5: Train Ridge Regression
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_train, y_train)

# Step 6: Predict on test set
y_pred_test = ridge_model.predict(X_test)

# Step 7: Evaluate using X_test and y_pred_test
r2 = r2_score(y_test, y_pred_test)
mse = mean_squared_error(y_test, y_pred_test)
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1 - ((1 - r2) * (n - 1)) / (n - p - 1)

# Step 8: Print results
print("Ridge R² Score on Test Set:", round(r2, 4))
print("Adjusted R² Score:", round(adjusted_r2, 4))
print("Mean Squared Error:", round(mse, 2))
```

```python
#8. If you divide tumors into small-radius and large-radius groups (using median split),how do their mean smoothness values differ?
import pandas as pd

# Find the median of mean radius
median_radius = df["mean radius"].median()

# Split into two groups
small_radius = df[df["mean radius"] <= median_radius]["mean smoothness"]
large_radius = df[df["mean radius"] > median_radius]["mean smoothness"]

# Calculate average smoothness for each group
print("Mean smoothness (small radius):", small_radius.mean())
print("Mean smoothness (large radius):", large_radius.mean())
```

```python
import matplotlib.pyplot as plt

df["radius_group"] = ["Small" if r <= median_radius else "Large" 
                      for r in df["mean radius"]]

df.boxplot(column="mean smoothness", by="radius_group")
plt.title("Mean Smoothness by Tumor Radius Group")
plt.suptitle("")
plt.show()
```

```python
# From your boxplot of mean smoothness by tumor radius group:

# The median smoothness (green line) for small-radius tumors looks slightly higher than for large-radius tumors.

# Both groups have a similar spread (IQR), but large-radius tumors show a few higher outliers.

# The difference isn’t huge, but it suggests small-radius tumors tend to have marginally smoother surfaces.
```

```python
#9. Which feature (out of mean concavity, mean concave points, and mean symmetry) shows the largest spread (variance)?
spread = df[["mean concavity", "mean concave points", "mean symmetry"]].var()
print(spread)
print("Feature with largest spread:", spread.idxmax())
```

```python
##10. If you want to predict tumor size, would you choose mean radius or mean area? Why?
# If your goal is predicting tumor size (volume/extent):

# Mean radius is the more fundamental measurement (a linear dimension).

# Mean area is derived from radius and thus redundant.

# In practice:

# Mean radius is usually chosen, because it is simpler and less noisy (area introduces squared variation).

# Using both may cause multicollinearity in models, since they are highly correlated.

# ✅ Conclusion

# 👉 To predict tumor size, you’d generally choose mean radius — it’s the base measurement, directly related to tumor geometry, and avoids redundancy with area.

# ⚡ Do you want me to also quantify their correlation (Pearson r) in your dataset to show how redundant they are?
```

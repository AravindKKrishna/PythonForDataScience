**class pandas.DataFrame(data=None, index=None, columns=None, dtype=None, copy=None)**

Two-dimensional, size-mutable, potentially heterogeneous
tabular data.

Data structure also contains labeled axes (rows and columns). Arithmetic operations align on both row and column labels.
>> contain same data type
>> input as list, ndarray, dict, scalar
>> data is in labed form index based along x-axis and y-axis

```python
#create a dataframe from list

import pandas as pd
data = {"number":[1,2,3,4,5],"name":['a','d','f','p','h']}
df = pd.DataFrame(data,index=[1,2,3,4,5],dtype=str)
df
```

```python
#create a dataframe from dict

data1={
       0:{"number":1,"name":'a'},
       1:{"number":2,"name":'b'},
       2:{"number":3,"name":'c'},
       3:{"number":4,"name":'d'},
       4:{"number":5,"name":'e'}
}

df1=pd.DataFrame.from_dict(data1)
print(df1)
```

```python
data = {
    0: {'Name': 'Alice', 'Age': 25},
    1: {'Name': 'Bob', 'Age': 30},
    2: {'Name': 'Charlie', 'Age': 35}
}

df = pd.DataFrame.from_dict(data, orient='index')
print(df)
```

```python
#create a dataframe from series

olimpyc_participation=pd.Series([205,204,201,200,197],index=[2012,2008,2004,2000,1996])
olimpyc_country=pd.Series(['london','beijing','athens','sydney','athens'],index=[2012,2008,2004,2000,1996])
df3=pd.DataFrame({'olimpyc_participation':olimpyc_participation,'olimpyc_country':olimpyc_country})
```

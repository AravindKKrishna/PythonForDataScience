```python
#calculating variance
glucose = df["Glucose"]

# Step 1: Calculate mean
mean_val = sum(glucose) / len(glucose)

# Step 2: Apply variance formula (sample variance: divide by n-1)
variance = sum((x - mean_val) ** 2 for x in glucose) / (len(glucose) - 1)

print("Mean of Glucose:", mean_val)
print("Variance of Glucose:", variance)
```

```python
import pandas as pd

# Example dataset
# df = pd.read_csv("your_dataset.csv")

# Split into younger and older groups
younger = df[df["Age"] < 30]
older   = df[df["Age"] >= 30]

# Calculate average BMI
avg_bmi_younger = younger["BMI"].mean()
avg_bmi_older   = older["BMI"].mean()

print("Average BMI of younger (<30):", avg_bmi_younger)
print("Average BMI of older (≥30):", avg_bmi_older)
# Create a column for age group
df["AgeGroup"] = df["Age"].apply(lambda x: "Younger (<30)" if x < 30 else "Older (≥30)")
```

```python
df.groupby("AgeGroup")["BMI"].mean().plot(kind="bar", color=["skyblue", "orange"])
plt.ylabel("Average BMI")
plt.legend()
plt.title("Average BMI of Younger vs Older Patients")
plt.show()
```

```python
#this shows age > 30 have high bmi
```

```python
import pandas as pd

# Example: df = pd.read_csv("your_dataset.csv")

# Define groups
normal_bmi = df[(df["BMI"] >= 18.5) & (df["BMI"] < 25)]
obese_bmi  = df[df["BMI"] >= 30]

# Calculate average SkinThickness
avg_skin_normal = normal_bmi["SkinThickness"].mean()
avg_skin_obese  = obese_bmi["SkinThickness"].mean()

print("Average SkinThickness (Normal BMI):", avg_skin_normal)
print("Average SkinThickness (Obese BMI):", avg_skin_obese)
import matplotlib.pyplot as plt

# Create BMI category column
def bmi_category(bmi):
    if 18.5 <= bmi < 25:
        return "Normal"
    elif bmi >= 30:
        return "Obese"
    else:
        return "Other"

df["BMI_Category"] = df["BMI"].apply(bmi_category)
```

```python
df.groupby("BMI_Category")["SkinThickness"].mean().loc[["Normal","Obese"]].plot(
    kind="bar", color=["green","red"]
)
plt.ylabel("Average SkinThickness")
plt.title("Average SkinThickness: Normal vs Obese BMI")
plt.show()
```

```python
#obese people have higher skin thicknes
```

```python
# Visualize the relationship in scatterplot in between the AveRooms and AveBedrms

sns.scatterplot(x='AveRooms', y='AveBedrms', data=df)
plt.title("AveRooms vs. AveBedrms")
plt.xlabel("Average Rooms")
plt.ylabel("Average Bedrooms")
plt.show()
```

```python
# scatter plot
sns.scatterplot(x="MedInc", y="MedHouseVal", data=df)
plt.xlabel("MedInc")
plt.ylabel("HousePrice")
plt.show()
```

It does show a clear upward trend between MedInc (median income) and MedHouseVal (median house value). As median income increases, house prices tend to rise as well. The points in the plot generally slope upward, indicating a positive correlation.

**This visual trend supports the earlier correlation result of approximately 0.688, which confirms a strong positive relationship between income and house prices. So yes—this code effectively reveals that trend.**

```python
# Compute the correlation matrix
corr_matrix = df.corr()

# Set up the matplotlib figure
plt.figure(figsize=(10, 8))

# Draw the heatmap
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm")

# Add title
plt.title("Feature Correlation Heatmap")

# Show the plot
plt.show()
```

Based on the correlation matrix and heatmap discussed on the page, the feature most strongly correlated with house prices (MedHouseVal) is MedInc, which stands for median income.

Correlation Value: Approximately 0.688, indicating a strong positive relationship.

```python
# Scatter plot: Population vs. Median House Value
plt.scatter(df['Population'], df['MedHouseVal'], alpha=0.5)

# Label axes
plt.xlabel('Population')
plt.ylabel('Median House Value')

# Add title
plt.title('Population vs. House Price')

# Show plot
plt.show()
```

Based on the scatter plot shown on the page—Population vs. Median House Value—there does not appear to be a strong visible relationship between the two variables.

What the plot suggests:

 The data points are widely scattered with no clear upward or downward trend.
 
 This indicates that Population does not have a significant linear correlation with house prices.

```python
# Divide income into 4 groups
df['IncomeGroup'] = pd.qcut(df['MedInc'], 4)

# Find average house price for each group
print(df.groupby('IncomeGroup')['MedHouseVal'].mean())
```

pd.qcut(df['MedInc'], 4) -- This line divides the MedInc column (which contains median income values) into 4 equal-sized groups based on income levels. These groups represent:

**Lowest income**

**Lower-middle income**

**Upper-middle income**

**Highest income**

Each house in the dataset is assigned to one of these four income groups.

```python
# Scatter plot: Population vs. House Price
plt.scatter(df['Population'], df['MedHouseVal'], alpha=0.5)
plt.xlabel('Population')
plt.ylabel('Median House Value')
plt.title('Population vs. House Price')
plt.show()
```

There is no visible relationship between population and house price.

High or low population areas can have both expensive and cheap houses, indicating weak or no correlation.

```python
# Scatter plot: Longitude vs. Latitude, colored by House Price
plt.scatter(df['Longitude'], df['Latitude'], c=df['MedHouseVal'], cmap='viridis', alpha=0.5)
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('House Prices by Location')
plt.colorbar(label='Median House Value')
plt.show()
```

There is a visible trend in house prices based on location.

Geographic factors (like proximity to coastlines or city centers) appear to affect house values, confirming a location effect.

```python
import numpy as np
from scipy.linalg import qr
from scipy.linalg import lu_factor

# Step 1: Keep only numeric columns from the DataFrame
numeric_df = df.select_dtypes(include=[np.number])

# Step 2: Convert to NumPy array
matrix = numeric_df.to_numpy()

# Step 3: QR decomposition with column pivoting
Q, R, pivot_columns = qr(matrix, pivoting=True)

# Step 4: Determine the rank of the matrix
rank = np.linalg.matrix_rank(matrix)

# Step 5: Extract basis columns using pivot indices
basis = matrix[:, pivot_columns[:rank]]

# Step 6: Print the basis
print("Basis for the column space:")
try:
    print("Basis for the column space:")
    print(basis)
except Exception as e:
    print("Error while printing basis:", e)

```

```python
a=pd.DataFrame(data=basis)
print(a)
```

```python
#multilinear regression
x=df[["MedInc","HouseAge","AveRooms","Latitude"]]
y=df["MedHouseVal"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

# Step 2: View predictions
print("Predicted values:")
y_pred=y_pred.reshape(-1,1)
print(y_pred)

c=pd.DataFrame(data=y_pred)

print(c)
```

```python
score = model.score(X_test, y_test)
print("Model R² score:", score)
```

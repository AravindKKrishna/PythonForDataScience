```python
# n=int(input("enter the number"))
def sumb(n):
    if n==1:
        return 1
    return n+sumb(n-1)
sumb(5)
```

```python
def recsum(n):
    if not n:
        return 0
    return n[0]+recsum(n[1:])
n=[2,4,3,5,6,7]
recsum(n)
```

```python
def recursive_list_sum(numbers):
    total = 0
    for element in numbers:
        if type(element)==list:  # If element is a list, recurse into it
            total += recursive_list_sum(element)
        else:
            total += element  # Otherwise add the number directly
    return total

# Example usage
numbers = [1, [2, [3, 4], 5], 6]
result = recursive_list_sum(numbers)
print("Sum of recursive list:", result)
```

```python
def fibinocii(n):
    if n==1:
        return 1
    if n==0:
        return 0
    return fibinocii(n-1)+fibinocii(n-2)
num = 10  
print("Fibonacci series:")
for i in range(num):
    print(fibinocii(i), end=" ")
```

```python
def sum_of_digits(n):
    if n==1:
        return 1
    return (n % 10) + sum_of_digits(n // 10)
sum_of_digits(12345)
```

```python
def sum_of_digits(n):
    # Base case: if n is a single digit, return n
    if n == 0:
        return 0
    else:
        # Get the last digit + recursive call on the rest
        return (n % 10) + sum_of_digits(n // 10)

# Example usage
num = int(input("Enter a number: "))
print("Sum of digits:", sum_of_digits(num))
```

```python
def sum_series(n):
    # Base case: if n <= 0, stop recursion
    if n <= 0:
        return 0
    else:
        return n + sum_series(n - 2)

# Test Data
print("sum_series(6) ->", sum_series(6))   # Expected: 12
print("sum_series(10) ->", sum_series(10)) # Expected: 30
```

```python
def harmonic_sum(n):
    # Base case
    if n == 1:
        return 1
    else:
        return 1/n + harmonic_sum(n-1)

# Example usage
n = int(input("Enter number of terms: "))
print(f"Harmonic sum up to {n} terms is:", harmonic_sum(n))
```

```python
def geometric_sum(a, r, n):
    # Base case
    if n == 0:
        return 0
    else:
        return a * (r ** (n - 1)) + geometric_sum(a, r, n - 1)

# Example usage
a = float(input("Enter first term (a): "))
r = float(input("Enter common ratio (r): "))
n = int(input("Enter number of terms (n): "))

print(f"Geometric sum up to {n} terms is:", geometric_sum(a, r, n))
```

```python
def power(a, b):
    # Base case
    if b == 0:
        return 1
    else:
        return a * power(a, b - 1)

# Test Data
print("power(3, 4) ->", power(3, 4))  # Expected: 81
```

```python
def gcd(a, b):
    # Base case
    if b == 0:
        return a
    else:
        return gcd(b, a % b)

# Example usage
x = int(input("Enter first number: "))
y = int(input("Enter second number: "))

print(f"GCD of {x} and {y} is:", gcd(x, y))
```

```python
import csv

def average_salary(filename):
    total_salary = 0
    count = 0

    with open(filename, "r") as file:
        reader = csv.DictReader(file)
        for row in reader:
            total_salary += int(row["Salary"])
            count += 1

    return total_salary / count if count > 0 else 0

# Example usage
filename = "employees.csv"
print("Average Salary:", average_salary(filename))
```

```python
class BankAccount:
    def __init__(self, account_number, name, balance=0.0):
        self.account_number = account_number
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount >= 0:
            if self.balance >= amount:
                self.balance -= amount
                print(f"₹{amount} withdrawn successfully.")
            else:
                print("Insufficient balance!")
        else:
            print("Withdrawal amount must be positive.")

    def display_balance(self):
        print(f"Account Number: {self.account_number}")
        print(f"Account Holder: {self.name}")
        print(f"Current Balance: ₹{self.balance}")

account1 = BankAccount(1001, "Aravind", 5000)


account1.display_balance()
account1.deposit(100)
account1.withdraw(3000)
account1.display_balance()
account1.withdraw(5000) 
```

```python
import math

# Parent class
class Shape:
    def area(self):
        return 0  # Default implementation

# Subclass Circle
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    # Override area()
    def area(self):
        return math.pi * (self.radius ** 2)

# Subclass Rectangle
class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    
    # Override area()
    def area(self):
        return self.length * self.width


# Example usage
circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Area of Circle:", circle.area())
print("Area of Rectangle:", rectangle.area())
```

```python
# Grocery Item Class
class GroceryItem:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    
    def get_discounted_price(self):
        # 10% discount for groceries
        return self.price * 0.90


# Electronic Item Class
class ElectronicItem:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    
    def get_discounted_price(self):
        # Flat 100 off if price > 1000, else no discount
        if self.price > 1000:
            return self.price - 100
        return self.price


# Polymorphic function
def calculate_total_price(item):
    return item.get_discounted_price()


# Example usage
apple = GroceryItem("Apple", 200)
tv = ElectronicItem("TV", 1500)

print(f"Total price for {apple.name}: {calculate_total_price(apple)}")
print(f"Total price for {tv.name}: {calculate_total_price(tv)}")
```

```python
# Base class
class Transport:
    def move(self):
        print("This transport can move.")


# Subclass Car
class Car(Transport):
    def move(self):
        print("Car drives on roads.")


# Subclass Bike
class Bike(Transport):
    def move(self):
        print("Bike rides on roads.")


# Subclass Airplane
class Airplane(Transport):
    def move(self):
        print("Airplane flies in the sky.")


# Create objects
car = Car()
bike = Bike()
plane = Airplane()

# Store them in a list
vehicles = [car, bike, plane]

# Demonstrate polymorphism
for v in vehicles:
    v.move()
```

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Less than (<)
    def __lt__(self, other):
        return self.marks < other.marks

    # Greater than (>)
    def __gt__(self, other):
        return self.marks > other.marks


# Example usage
s1 = Student("Aravind", 85)
s2 = Student("Ashkar", 92)

print(s1 < s2)   # True (85 < 92)
print(s1 > s2)   # False (85 > 92)
```

```python
class Appliance:
    def __init__(self, brand, power):
        self.brand = brand
        self.power = power


class WashingMachine(Appliance):
    def __init__(self, brand, power, wash_capacity):
        super().__init__(brand, power)   # call parent constructor
        self.wash_capacity = wash_capacity

    def info(self):
        print(f"The brand of washing machine is {self.brand}, "
              f"its power is {self.power} W, "
              f"and its wash capacity is {self.wash_capacity} kg.")


class Refrigerator(Appliance):
    def __init__(self, brand, power, cooling_capacity):
        super().__init__(brand, power)   # inherit from Appliance
        self.cooling_capacity = cooling_capacity

    def info(self):
        print(f"The brand of refrigerator is {self.brand}, "
              f"its power is {self.power} W, "
              f"and its cooling capacity is {self.cooling_capacity} L.")


# Example usage
wm = WashingMachine("LG", 2000, 7)
fridge = Refrigerator("Samsung", 150, 250)

wm.info()
fridge.info()
```

```python
# Base class
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def calculate_bonus(self):
        # Base version (can be abstract in real-world use)
        return 0


# Subclass: Manager
class Manager(Employee):
    def calculate_bonus(self):
        # Managers get 20% bonus
        return self.salary * 0.20


# Subclass: Developer
class Developer(Employee):
    def calculate_bonus(self):
        # Developers get 10% bonus
        return self.salary * 0.10


# Example usage
emp1 = Manager("Aravind", 50000)
emp2 = Developer("Ashkar", 40000)

print(f"{emp1.name} (Manager) Bonus: {emp1.calculate_bonus()}")
print(f"{emp2.name} (Developer) Bonus: {emp2.calculate_bonus()}")
```

```python
from abc import ABC, abstractmethod

# Abstract class
class Transport(ABC):
    def __init__(self, distance):
        self.distance = distance

    @abstractmethod
    def fare(self):
        pass


# Subclass: Bus
class Bus(Transport):
    def __init__(self, distance, rate_per_km=5):
        super().__init__(distance)
        self.rate_per_km = rate_per_km

    def fare(self):
        return self.distance * self.rate_per_km


# Subclass: Train
class Train(Transport):
    def __init__(self, distance, rate_per_km=3):
        super().__init__(distance)
        self.rate_per_km = rate_per_km

    def fare(self):
        return self.distance * self.rate_per_km


# Example usage
bus = Bus(100)      # 100 km
train = Train(200)  # 200 km

print(f"Bus Fare for 100 km: ₹{bus.fare()}")
print(f"Train Fare for 200 km: ₹{train.fare()}")
```
